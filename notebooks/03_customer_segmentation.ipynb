# Notebook 03: Customer Segmentation using K-Means Clustering

## 📌 Objectives
Group customer base into 3 distinct behavioral segments:
1. **At-Risk Customers** (High churn rate, high support tickets, low logins)
2. **Loyal Power Users** (Long tenure, low churn rate, high usage)
3. **New / Casual Users** (Moderate tenure, balanced usage)


```python
import pandas as pd
import numpy as np
import joblib

df_clean = pd.read_csv('../data/processed/cleaned_churn_data.csv')
kmeans = joblib.load('../models/kmeans_cluster_model.joblib')
cluster_scaler = joblib.load('../models/cluster_scaler.joblib')

cluster_features = ["Tenure", "Monthly Charges", "Support Tickets", "Login Frequency", "Usage Hours"]
scaled_feats = cluster_scaler.transform(df_clean[cluster_features])
df_clean['ClusterID'] = kmeans.predict(scaled_feats)

df_clean.groupby('ClusterID')[cluster_features].mean()
```
