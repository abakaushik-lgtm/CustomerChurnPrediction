# Notebook 01: Customer Churn Exploratory Data Analysis (EDA)

## 📌 Executive Overview
This notebook performs comprehensive Exploratory Data Analysis on the **Customer Churn Dataset** (7,043 rows, 20 features).
We investigate key demographic features, billing contract types, monthly charges, support tickets, and login frequency to identify churn drivers.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')

# Load raw dataset
df = pd.read_csv('../data/raw/customer_churn_data.csv')
print("Dataset Shape:", df.shape)
df.head()
```

## 📊 1. Missing Values & Summary Statistics

```python
print("Missing Values per Column:")
print(df.isna().sum())

print("
Summary Statistics:")
df.describe()
```

## 📉 2. Churn Class Distribution

```python
churn_counts = df['Churn'].value_counts()
churn_rate = (df['Churn'] == 'Yes').mean() * 100

print(f"Overall Churn Rate: {churn_rate:.2f}%")

plt.figure(figsize=(6, 4))
sns.countplot(x='Churn', data=df, palette=['#10B981', '#F43F5E'])
plt.title('Customer Churn Distribution', fontsize=14, fontweight='bold')
plt.ylabel('Customer Count')
plt.show()
```

## 📑 3. Churn Rate by Contract Type & Subscription

```python
plt.figure(figsize=(10, 4))
sns.barplot(x='Contract Type', y='Monthly Charges', hue='Churn', data=df, palette=['#38BDF8', '#F43F5E'])
plt.title('Monthly Charges by Contract Type & Churn Status', fontsize=14, fontweight='bold')
plt.show()

contract_churn = df.groupby('Contract Type')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100)
print("Churn Rate by Contract Type (%):")
print(contract_churn)
```

## 💡 Key Business Takeaways
1. **Contract Type Impact**: Customers on Month-to-Month contracts exhibit the highest churn rate (~42.7%).
2. **Tenure Loyalty**: Customers with tenure > 24 months show less than 10% churn probability.
3. **Monthly Charges Sensitivity**: Customers paying over $70/month churn at nearly double the rate of low-tier users.
